```python
from sklearn.tree import DecisionTreeClassifier

# Input features:
# [Income (in lakhs), Credit Score, Loan Amount (in lakhs)]
X = [
    [8, 750, 5],
    [5, 650, 8],
    [12, 800, 6],
    [4, 600, 10],
    [10, 780, 5],
    [3, 580, 7],
    [15, 820, 8],
    [6, 700, 4],
    [9, 760, 6],
    [2, 550, 9]
]

# Output: Loan decision
y = [
    "Approved",
    "Rejected",
    "Approved",
    "Rejected",
    "Approved",
    "Rejected",
    "Approved",
    "Approved",
    "Approved",
    "Rejected"
]

# Create Decision Tree
model = DecisionTreeClassifier(
    criterion="entropy",
    max_depth=3
)

# Train the model
model.fit(X, y)

# New customer
# Income = 10 lakhs
# Credit Score = 770
# Loan Amount = 5 lakhs
new_customer = [[10, 770, 5]]

# Prediction
prediction = model.predict(new_customer)

print("Loan Decision:", prediction[0])
```
