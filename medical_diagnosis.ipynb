```python
# Medical Diagnosis

symptoms = {
    "fever": False,
    "cough": False,
    "body_pain": False,
    "sore_throat": False
}

# Input symptoms
for symptom in symptoms:
    answer = input("Do you have " + symptom + "? (yes/no): ")
    symptoms[symptom] = answer.lower() == "yes"

# Diagnosis rules
if symptoms["fever"] and symptoms["body_pain"]:
    print("\nDiagnosis: Flu")
elif symptoms["cough"] and symptoms["sore_throat"]:
    print("\nDiagnosis: Common Cold")
elif symptoms["fever"] and symptoms["cough"]:
    print("\nDiagnosis: Possible Infection")
else:
    print("\nDiagnosis: No clear diagnosis")
```
