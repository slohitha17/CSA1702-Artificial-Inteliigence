```python
# Family Tree

father = {
    "arun": "kumar",
    "ravi": "arun"
}

mother = {
    "arun": "meena",
    "ravi": "sita"
}

def get_father(person):
    if person in father:
        print(person, "is the child of", father[person])

def get_mother(person):
    if person in mother:
        print(person, "is the child of", mother[person])

person = input("Enter person's name: ").lower()

get_father(person)
get_mother(person)
```
