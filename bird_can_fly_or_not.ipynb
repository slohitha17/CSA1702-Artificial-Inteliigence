```python
# Bird Can Fly or Not

birds = {
    "eagle": True,
    "sparrow": True,
    "parrot": True,
    "penguin": False,
    "ostrich": False
}

bird = input("Enter bird name: ").lower()

if bird in birds:
    if birds[bird]:
        print(bird, "can fly.")
    else:
        print(bird, "cannot fly.")
else:
    print("Bird not found.")
```
