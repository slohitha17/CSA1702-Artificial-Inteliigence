```python
# Backward Chaining

facts = {
    "rain"
}

rules = [
    ("wet_ground", ["rain"]),
    ("slippery", ["wet_ground"]),
    ("dangerous", ["slippery"])
]

def backward_chaining(goal, facts, rules):

    # If goal is already a fact
    if goal in facts:
        return True

    # Find a rule that produces the goal
    for conclusion, conditions in rules:
        if conclusion == goal:

            print("Checking:", goal)

            # Check all conditions
            for condition in conditions:
                if not backward_chaining(condition, facts, rules):
                    return False

            print("Proved:", goal)
            return True

    return False


# Query
goal = "dangerous"

if backward_chaining(goal, facts, rules):
    print("\nGoal", goal, "is TRUE")
else:
    print("\nGoal", goal, "is FALSE")
```
