```python
# Fruit and its Color

fruits = [
    ("apple", "red"),
    ("banana", "yellow"),
    ("orange", "orange"),
    ("grapes", "green"),
    ("mango", "yellow"),
    ("watermelon", "green")
]

fruit = input("Enter fruit: ").lower()

found = False

for name, color in fruits:
    if name == fruit:
        print("Color =", color)
        found = True

if not found:
    print("Fruit not found.")
```
