```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Create dataset
data = {
    'CreditScore': [750,650,720,680,800,690,710,600,780,670],
    'Income': [60000,40000,55000,45000,70000,48000,52000,35000,65000,42000],
    'Result': ['Approve','Reject','Approve','Reject','Approve',
               'Reject','Approve','Reject','Approve','Reject']
}

df = pd.DataFrame(data)

# Input features
X = df[['CreditScore', 'Income']]

# Target
y = df['Result']

# Create Decision Tree
model = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=2,
    random_state=42
)

# Train model
model.fit(X, y)

# Predict a new applicant
new_applicant = [[720, 55000]]

prediction = model.predict(new_applicant)

print("Predicted Result:", prediction[0])

# Display tree
plt.figure(figsize=(12, 7))
plot_tree(
    model,
    feature_names=X.columns,
    class_names=model.classes_,
    filled=True
)
plt.show()
```
